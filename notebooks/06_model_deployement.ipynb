## Notebook 06 - Model Deployment & Prediction
### Objective

The goal of this notebook is to load the trained Random Forest model and use it to predict hourly taxi demand for new input data. This notebook demonstrates how the trained model can be deployed and integrated into an application such as a dashboard or API.

````python
import pandas as pd
import joblib

## pandas → Used to create input data in DataFrame format.
## joblib → Used to load the saved machine learning model (best_model.pkl)
````

````python
# Load the best trained model

model = joblib.load("../models/best_model.pkl")

print("Model loaded successfully.")
````

````python
# Create sample input for prediction

sample_input = pd.DataFrame({
    "PULocationID": [161],
    "hour": [10],
    "day_of_week": [2],
    "month": [3],
    "is_weekend": [0],
    "lag_1": [42],
    "lag_24": [36],
    "rolling_mean_3": [39.3],
    "rolling_mean_24": [37.8],
    "rolling_std_24": [4.2],
    "is_peak_hour": [1]
})

sample_input
````

````python
# Predict hourly taxi demand

predicted_demand = model.predict(sample_input)

print(f"Predicted Taxi Demand: {predicted_demand[0]:.2f}")
````

````python
# Display input features along with prediction

result = sample_input.copy()
result["Predicted_Demand"] = predicted_demand

result
````

````python
# Save prediction results

result.to_csv("../reports/sample_prediction.csv", index=False)

print("Prediction results saved successfully.")
````

# Streamlit vs FastAPI

Both **Streamlit** and **FastAPI** can use the same trained Machine Learning model (`best_model.pkl`), but they serve different purposes.

---

# Streamlit

Streamlit is primarily a **frontend framework** used to build interactive Machine Learning applications and dashboards.

A Streamlit application can:

- Load the trained model.
- Accept user inputs through a graphical interface.
- Generate predictions.
- Display results using tables, charts, and visualizations.

Architecture:

```
User
   │
   ▼
Streamlit Dashboard
   │
   ▼
Load best_model.pkl
   │
   ▼
Random Forest Model
   │
   ▼
Prediction
   │
   ▼
Display Result
```

A Streamlit application can be deployed to the cloud (e.g., Streamlit Community Cloud or Render), allowing multiple users to access it through a web browser.

However, the Machine Learning model remains embedded inside the Streamlit application. If multiple applications (such as a website, Android app, and iOS app) all require predictions, each application would need its own prediction logic or backend.

For this reason, Streamlit is commonly used for:

- Machine Learning dashboards
- Portfolio projects
- Internal company tools
- Demonstrations and prototypes

---

# FastAPI

FastAPI is a **backend framework** used to expose the Machine Learning model as an API.

Instead of displaying a user interface, FastAPI waits for requests from different applications, performs the prediction, and returns the result.

Architecture:

```
Website
      │
Mobile App
      │
Desktop App
      │
      ▼
     FastAPI
      │
      ▼
Load best_model.pkl
      │
      ▼
Random Forest Model
      │
      ▼
Prediction
      │
      ▼
Return Response
```

In this architecture:

- The model exists only once.
- Every application sends requests to the same backend.
- The backend performs the prediction and returns the result.

This approach is scalable and is commonly used in production systems.

---

# Why Use FastAPI?

Suppose FlowCast is available on:

- A Website
- An Android App
- An iPhone App

Without FastAPI:

```
Website
   │
   ▼
Model

Android
   │
   ▼
Model

iPhone
   │
   ▼
Model
```

Each application would require its own prediction logic.

With FastAPI:

```
Website ───┐
           │
Android ───┼────► FastAPI
           │
iPhone ────┘
                 │
                 ▼
        Random Forest Model
```

All applications share the same backend and the same trained model.

# What is Vercel?

Vercel is **not a frontend or backend framework**.

It is a **cloud hosting platform**.

Its job is simply to host applications on the internet.

Examples:

| Technology | Purpose |
|------------|---------|
| Streamlit | Frontend / Dashboard |
| FastAPI | Backend / API |
| Vercel | Hosting Platform |
| Render | Hosting Platform |
| AWS | Cloud Hosting |
| Azure | Cloud Hosting |

---

# Can Streamlit Be Hosted on Vercel?

Generally, **no**.

Vercel is primarily designed for:

- React
- Next.js
- Frontend web applications
- Serverless functions

Streamlit applications are typically deployed using:

- Streamlit Community Cloud
- Render

FastAPI applications are commonly deployed on:

- Render

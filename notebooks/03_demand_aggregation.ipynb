````python
df = pd.read_parquet("../data/processed/demand_data.parquet")

````

````python
df["pickup_hour_timestamp"] = df["tpep_pickup_datetime"].dt.floor("h")

````

````python
demand_df = (df.groupby(["PULocationID", "pickup_hour_timestamp"]).size().reset_index(name="demand"))
````

# 03. Demand Aggregation

## Objectives
- Transform trip-level data into an hourly demand forecasting dataset.
- Create the target variable (`demand`) for machine learning.
- Generate basic time-based features.
- Save the aggregated dataset for feature engineering.

---

## Steps Performed

### 1. Loaded Processed Data
- Imported the cleaned demand dataset containing:
  - `PULocationID`
  - `tpep_pickup_datetime`

### 2. Rounded Pickup Time
- Rounded each pickup timestamp to the nearest hour using:

```python
dt.floor("h")
```

- Created:
  - `pickup_hour_timestamp`

This groups all trips occurring within the same hour.

---

### 3. Created Target Variable

Grouped data by:

- Pickup Location ID
- Pickup Hour

Then counted the number of trips.

```python
groupby(["PULocationID", "pickup_hour_timestamp"]).size()
```

Created:

- `demand`

This represents:

> Number of taxi pickups in a particular zone during a specific hour.

This is the prediction target for the ML model.

---

### 4. Created Time Features

Extracted:

- Date
- Hour
- Day of Week
- Month
- Weekend Indicator

These features help the model learn temporal demand patterns.

---

### 5. Saved Aggregated Dataset

Saved the final dataset as:

```
data/processed/hourly_demand.parquet
```

---

## Output

A structured hourly demand dataset where each row represents:

- Pickup Zone
- Pickup Hour
- Hourly Demand
- Time-based Features

This dataset is now ready for advanced feature engineering and machine learning model development.

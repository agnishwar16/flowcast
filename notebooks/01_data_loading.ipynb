```python
# Remove duplicate rows (if any)
df = df.drop_duplicates()
```

```python
# Convert pickup and dropoff to datetime
df["tpep_pickup_datetime"] = pd.to_datetime(df["tpep_pickup_datetime"])
df["tpep_dropoff_datetime"] = pd.to_datetime(df["tpep_dropoff_datetime"])
```

```python
df.columns
```

```python
# Missing values
df.isnull().sum()
```

```python
# Remove rows with missing pickup/dropoff times
df = df.dropna(subset=["tpep_pickup_datetime", "tpep_dropoff_datetime"])
```

```python
# Remove invalid passenger counts
df = df[df["passenger_count"] > 0]
```

```python
# Remove invalid trip distances
df = df[df["trip_distance"] > 0]
```

```python
# Remove trips with non-positive fares
df = df[df["fare_amount"] > 0]
```

```python
# Remove trips where dropoff is before pickup
df = df[df["tpep_dropoff_datetime"] > df["tpep_pickup_datetime"]]
```

```python
# Check remaining shape
df.shape
```

```python
df.isnull().sum()
```

FEATURE ENG

```python
# Trip duration (minutes)
df["trip_duration_min"] = (
    df["tpep_dropoff_datetime"] - df["tpep_pickup_datetime"]
).dt.total_seconds() / 60
```

```python
# Remove unrealistic durations
df = df[(df["trip_duration_min"] > 0) & (df["trip_duration_min"] <= 180)]
```

```python
# Extract time features
df["pickup_hour"] = df["tpep_pickup_datetime"].dt.hour
df["pickup_day"] = df["tpep_pickup_datetime"].dt.day
df["pickup_month"] = df["tpep_pickup_datetime"].dt.month
df["pickup_dayofweek"] = df["tpep_pickup_datetime"].dt.dayofweek
```

```python
# Weekend flag
df["is_weekend"] = df["pickup_dayofweek"].isin([5, 6]).astype(int)
```

# 01. Data Loading & Initial Preprocessing

## Objectives
- Load all monthly NYC Yellow Taxi trip datasets.
- Merge them into a single DataFrame.
- Perform initial data inspection.
- Clean invalid and inconsistent records.
- Create basic time-based features.
- Save the cleaned dataset for later stages.

---

## Steps Performed

### 1. Imported Libraries
- pandas
- pathlib

### 2. Located Dataset Files
- Used `Path` and `glob()` to automatically detect all monthly Parquet files.

### 3. Loaded and Merged Data
- Read all Parquet files.
- Combined them into a single DataFrame.

### 4. Initial Inspection
- Dataset shape
- First few rows
- Data types
- Summary statistics
- Missing values
- Duplicate rows
- Memory usage
- Column names

### 5. Data Cleaning
- Removed duplicate records.
- Removed rows with missing pickup/dropoff timestamps.
- Removed trips with:
  - Passenger count ≤ 0
  - Trip distance ≤ 0
  - Fare amount ≤ 0
  - Dropoff time before pickup time

### 6. Feature Engineering
Created:
- Trip Duration (minutes)
- Pickup Hour
- Pickup Day
- Pickup Month
- Pickup Day of Week
- Weekend Indicator

### 7. Saved Processed Data
- Exported the cleaned dataset as:
  `data/processed/cleaned_taxi_data.parquet`

---

## Output

A cleaned taxi trip dataset ready for Exploratory Data Analysis (EDA).
